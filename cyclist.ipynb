```python
all_data = pd.concat([df1,df2,df3,df4,df5,df6,df7,df8,df9,df10,df11,df12], ignore_index=True)
```

```python
all_data.head()
```

```python
all_data = all_data.drop(columns=['start_lat', 'start_lng', 'end_lat', 'end_lng'], errors='ignore')
```

```python
print("\nHow many rows are in data frame?", len(all_data))
print("\nDimensions of the data frame:", all_data.shape)
print("\nSee list of columns and data types:\n")
all_data.info()
```

```python
# Statistical summary of data (mainly for numerics)
print("\nStatistical summary of data:\n", all_data.describe())
```

```python
print("\nValue counts for 'member_casual' column after cleaning:\n",
all_data['member_casual'].value_counts())
```

```python
# Convert 'started_at' and 'ended_at' to datetime objects
all_data['started_at'] = pd.to_datetime(all_data['started_at'])
all_data['ended_at'] = pd.to_datetime(all_data['ended_at'])
```

```python
# Add columns that list the date, month, day, and year of each ride
all_data['date'] = all_data['started_at'].dt.date
all_data['month'] = all_data['started_at'].dt.month
all_data['day'] = all_data['started_at'].dt.day
all_data['year'] = all_data['started_at'].dt.year
all_data['day_of_week'] = all_data['started_at'].dt.day_name()
```

```python
# Add a "ride_length" calculation to all_trips (in seconds)
all_data['ride_length'] = (all_data['ended_at'] - all_data['started_at']).dt.total_seconds()
```

```python
all_trip = all_data[all_data["ride_length"]>=0].copy()
```

```python
#==============================
# CONDUCT DESCRIPTIVE ANALYSIS
#==============================

# Descriptive analysis on ride_length (all figures in seconds)
print("\nDescriptive statistics for ride_length:\n", all_trip['ride_length'].describe())

print("-------------------------------------------------------------------------")

# Compare members and casual users
print("\nRide length statistics grouped by member_casual:\n",
all_trip.groupby('member_casual')['ride_length'].agg(['mean', 'median', 'max', 'min']))


```

```python
# Order the days of the week for correct sorting for analysis
days_order = ["Sunday", "Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday"]
all_trip['day_of_week'] = pd.Categorical(all_trip['day_of_week'],
categories=days_order, ordered=True)

# See the average ride time by each day for members vs casual users
print("\nAverage ride length by member type and day of the week:\n",
all_trip.groupby(['member_casual', 'day_of_week'])['ride_length'].mean())

```

```python

# Analyze ridership data by type and weekday
summary_stats = all_trip.groupby(['member_casual', 'day_of_week']).agg(
number_of_rides=('ride_id', 'count'),
average_duration=('ride_length', 'mean')
).reset_index()

print("\nSummary of rides and duration by rider type and weekday:\n", summary_stats)
```

```python
#==========================================
# EXPORT SUMMARY FILE FOR FURTHER ANALYSIdS
#==========================================

# Create a .csv file that you will visualize elsewhere
import pandas as pd

# 1. Base aggregations (ride count & average duration)
summary = all_trip.groupby(['member_casual', 'day_of_week']).agg(
    number_of_rides=('ride_id', 'count'),
    average_duration=('ride_length', 'mean')
)

# 2. Pivot the rideable_type counts
bike_counts = all_trip.pivot_table(
    index=['member_casual', 'day_of_week'],
    columns='rideable_type',
    values='ride_id',
    aggfunc='count',
    fill_value=0
)

# 3. Combine both and export
counts = pd.concat([summary, bike_counts], axis=1).reset_index()
counts.to_csv('avg_ride_length.csv', index=False)
```
